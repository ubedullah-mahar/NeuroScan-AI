```python
# Original BRISC2025 dataset location
DATASET_ROOT = Path(r"D:\brain tumour\brisc2025")

CLASSIFICATION_ROOT = DATASET_ROOT / "classification_task"

TRAIN_DIR = CLASSIFICATION_ROOT / "train"
TEST_DIR = CLASSIFICATION_ROOT / "test"

# Four classification categories
CLASSES = [
    "glioma",
    "meningioma",
    "no_tumor",
    "pituitary"
]

# Verify paths
print("Dataset exists:", DATASET_ROOT.exists())
print("Classification directory exists:", CLASSIFICATION_ROOT.exists())
print("Training directory exists:", TRAIN_DIR.exists())
print("Testing directory exists:", TEST_DIR.exists())
```

```python
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

def get_image_files(directory):
    return [
        path for path in directory.rglob("*")
        if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
    ]


dataset_summary = []

for split_name, split_dir in [
    ("Train", TRAIN_DIR),
    ("Test", TEST_DIR)
]:
    for class_name in CLASSES:

        class_dir = split_dir / class_name
        images = get_image_files(class_dir)

        dataset_summary.append({
            "Split": split_name,
            "Class": class_name,
            "Image_Count": len(images)
        })


summary_df = pd.DataFrame(dataset_summary)

print(summary_df.to_string(index=False))

print("\nTotal images:", summary_df["Image_Count"].sum())
```

```python
import hashlib
from collections import defaultdict

def calculate_image_hash(image_path):

    with Image.open(image_path) as image:

        # Standardize representation
        grayscale = image.convert("L")

        # Include dimensions to distinguish differently sized images
        image_bytes = (
            str(grayscale.size).encode()
            + grayscale.tobytes()
        )

        return hashlib.sha256(image_bytes).hexdigest()


duplicate_groups = defaultdict(list)

for split_name, split_dir in [
    ("Train", TRAIN_DIR),
    ("Test", TEST_DIR)
]:

    for class_name in CLASSES:

        image_paths = get_image_files(
            split_dir / class_name
        )

        for image_path in image_paths:

            image_hash = calculate_image_hash(image_path)

            duplicate_groups[image_hash].append({
                "Split": split_name,
                "Class": class_name,
                "Filename": image_path.name,
                "Path": str(image_path)
            })


duplicates = {
    image_hash: records
    for image_hash, records in duplicate_groups.items()
    if len(records) > 1
}

print("Total unique image hashes:", len(duplicate_groups))
print("Duplicate image groups:", len(duplicates))

print(
    "Total images belonging to duplicate groups:",
    sum(len(records) for records in duplicates.values())
)
```

## 7. Near-Duplicate Image Detection

This experiment uses perceptual hashing to identify visually
similar MRI images across training and testing datasets.

Exact duplicate images are also identified separately so
that near-duplicate candidates can be investigated.

```python
import imagehash

# Store image metadata and perceptual hashes
phash_records = []

for split_name, split_dir in [
    ("Train", TRAIN_DIR),
    ("Test", TEST_DIR)
]:

    for class_name in CLASSES:

        for image_path in get_image_files(split_dir / class_name):

            with Image.open(image_path) as image:

                # Standardize image representation
                grayscale = image.convert("L")

                # Calculate perceptual hash
                phash = imagehash.phash(grayscale)

                phash_records.append({
                    "Split": split_name,
                    "Class": class_name,
                    "Filename": image_path.name,
                    "Path": str(image_path),
                    "pHash": phash
                })

phash_df = pd.DataFrame(phash_records)

print("Images processed:", len(phash_df))
print("Perceptual hashes generated successfully.")
```

```python
# Step 3: Optimized cross-split perceptual hash comparison

import numpy as np

# Convert perceptual hashes into arrays of bytes
train_hashes = phash_df[
    phash_df["Split"] == "Train"
].reset_index(drop=True)

test_hashes = phash_df[
    phash_df["Split"] == "Test"
].reset_index(drop=True)


def hashes_to_array(hash_series):
    return np.array([
        list(bytes.fromhex(str(h)))
        for h in hash_series
    ], dtype=np.uint8)


train_hash_array = hashes_to_array(train_hashes["pHash"])
test_hash_array = hashes_to_array(test_hashes["pHash"])


# Lookup table for counting set bits
bit_count_lookup = np.array(
    [bin(i).count("1") for i in range(256)],
    dtype=np.uint8
)


near_duplicate_candidates = []

batch_size = 64

for start in range(0, len(test_hash_array), batch_size):

    end = min(start + batch_size, len(test_hash_array))

    # XOR compares all hashes in this batch with training hashes
    xor_result = (
        test_hash_array[start:end, None, :]
        ^ train_hash_array[None, :, :]
    )

    # Calculate Hamming distances using vectorized operations
    distances = bit_count_lookup[xor_result].sum(axis=2)

    # Exclude exact matches (distance 0)
    # They have already been investigated using SHA-256
    rows, cols = np.where(
        (distances >= 1) & (distances <= 4)
    )

    for row, col in zip(rows, cols):

        test_index = start + row

        near_duplicate_candidates.append({
            "Test_Class": test_hashes.iloc[test_index]["Class"],
            "Train_Class": train_hashes.iloc[col]["Class"],
            "Test_Filename": test_hashes.iloc[test_index]["Filename"],
            "Train_Filename": train_hashes.iloc[col]["Filename"],
            "Hamming_Distance": int(distances[row, col])
        })

    print(f"Processed {end}/{len(test_hash_array)} test images")


near_duplicates_df = pd.DataFrame(near_duplicate_candidates)

print("\nNear-duplicate candidate pairs:", len(near_duplicates_df))

if not near_duplicates_df.empty:
    print(
        near_duplicates_df
        .sort_values("Hamming_Distance")
        .head(20)
        .to_string(index=False)
    )
```

```python
# Create a detailed audit of exact duplicate images

exact_duplicate_records = []

for image_hash, records in duplicates.items():

    splits = {record["Split"] for record in records}
    classes = {record["Class"] for record in records}

    if splits == {"Train", "Test"}:
        duplicate_type = "Train-Test"
    elif splits == {"Train"}:
        duplicate_type = "Train-Only"
    else:
        duplicate_type = "Test-Only"

    for record in records:

        exact_duplicate_records.append({
            "Hash": image_hash,
            "Duplicate_Type": duplicate_type,
            "Split": record["Split"],
            "Class": record["Class"],
            "Filename": record["Filename"],
            "Path": record["Path"],
            "Group_Size": len(records),
            "Class_Conflict": len(classes) > 1
        })

exact_duplicate_audit = pd.DataFrame(exact_duplicate_records)

# Summary
print("EXACT DUPLICATE AUDIT")
print("=" * 60)

print("\nDuplicate group counts:")
print(
    exact_duplicate_audit
    .groupby("Duplicate_Type")["Hash"]
    .nunique()
)

print("\nFiles belonging to duplicate groups:")
print(
    exact_duplicate_audit
    .groupby("Duplicate_Type")
    .size()
)

print("\nGroups with conflicting class labels:")
conflicting_groups = (
    exact_duplicate_audit[
        exact_duplicate_audit["Class_Conflict"]
    ]
    .groupby("Hash")
)

print("Number of conflicting groups:", len(conflicting_groups))

# Display all train-test duplicate records
print("\nTRAIN-TEST EXACT DUPLICATES")
print("=" * 60)

cross_split_exact_audit = exact_duplicate_audit[
    exact_duplicate_audit["Duplicate_Type"] == "Train-Test"
]

display(
    cross_split_exact_audit[
        ["Hash", "Split", "Class", "Filename", "Group_Size"]
    ].sort_values(["Hash", "Split"])
)
```

```python
# Create clean training and testing image path lists

train_paths_original = []

for class_name in CLASSES:
    train_paths_original.extend(
        get_image_files(TRAIN_DIR / class_name)
    )

test_paths_original = []

for class_name in CLASSES:
    test_paths_original.extend(
        get_image_files(TEST_DIR / class_name)
    )


# Remove training images that have exact copies in the test set

train_paths_clean = [
    path for path in train_paths_original
    if str(path) not in set(train_images_to_remove)
]

# Keep test set completely unchanged

test_paths_clean = test_paths_original.copy()


print("DATASET CLEANING SUMMARY")
print("=" * 60)

print("Original training images:", len(train_paths_original))
print("Training images removed:", len(train_paths_original) - len(train_paths_clean))
print("Clean training images:", len(train_paths_clean))
print("Test images:", len(test_paths_clean))

print("\nTest dataset unchanged:",
      len(test_paths_clean) == len(test_paths_original))
```

```python
import hashlib

def calculate_sha256(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as file:
        for chunk in iter(lambda: file.read(4096), b""):
            sha256.update(chunk)

    return sha256.hexdigest()


# Calculate hashes for cleaned training images
train_clean_hashes = {
    calculate_sha256(path)
    for path in train_paths_clean
}

# Calculate hashes for test images
test_hashes = {
    calculate_sha256(path)
    for path in test_paths_clean
}

# Find any remaining exact duplicates
remaining_leakage = train_clean_hashes.intersection(test_hashes)

print("LEAKAGE VERIFICATION")
print("=" * 60)

print("Unique cleaned training images:", len(train_clean_hashes))
print("Unique test images:", len(test_hashes))
print("Remaining exact duplicates:", len(remaining_leakage))

if len(remaining_leakage) == 0:
    print("\nSUCCESS: No exact duplicates remain between train and test.")
else:
    print("\nWARNING: Exact duplicates still exist!")
```

```python
from collections import defaultdict

def find_duplicate_groups(paths):
    hash_to_paths = defaultdict(list)

    for path in paths:
        image_hash = calculate_sha256(path)
        hash_to_paths[image_hash].append(path)

    duplicate_groups = {
        image_hash: group_paths
        for image_hash, group_paths in hash_to_paths.items()
        if len(group_paths) > 1
    }

    return duplicate_groups


# Find duplicates within cleaned training data
train_duplicates_clean = find_duplicate_groups(train_paths_clean)

# Find duplicates within test data
test_duplicates = find_duplicate_groups(test_paths_clean)


print("WITHIN-SPLIT DUPLICATE AUDIT")
print("=" * 60)

print("Training duplicate groups:", len(train_duplicates_clean))
print("Test duplicate groups:", len(test_duplicates))

print("\nDuplicate files in training groups:",
      sum(len(paths) for paths in train_duplicates_clean.values()))

print("Duplicate files in test groups:",
      sum(len(paths) for paths in test_duplicates.values()))


# Display training duplicate groups
print("\nTRAINING DUPLICATE GROUPS")
print("=" * 60)

for group_number, (image_hash, paths) in enumerate(
    train_duplicates_clean.items(), start=1
):
    print(f"\nGroup {group_number} ({len(paths)} copies):")

    for path in paths:
        print("  ", path)


# Display test duplicate groups
print("\nTEST DUPLICATE GROUPS")
print("=" * 60)

for group_number, (image_hash, paths) in enumerate(
    test_duplicates.items(), start=1
):
    print(f"\nGroup {group_number} ({len(paths)} copies):")

    for path in paths:
        print("  ", path)
```
